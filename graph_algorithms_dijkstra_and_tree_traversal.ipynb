```python
def my_dijkstra(node_l, node_d, start: str, end: str):
  Route = []
  while True:
    min = np.inf
    min_index = 0
    (x,y) = (0,0)
    #2
    if not(node_l):
      ##print("END")
      print(node_d)
      ans = [end]
      forward = ""
      while forward != start:
        for route in Route:
          if route[1] == end:
            forward = route[0]
            ans.insert(0, forward)
            end = forward
            break
      print(ans)
      break

    for i in node_l:
      if min >= node_l[i]:
        min = node_l[i]
        min_index = i

    node_d[min_index] = node_l[min_index]

    for i in range(len(edge[min_index].keys())):
      edge_index = list(edge[min_index].keys())[i]
      if (edge_index in node_d.keys()) and edge[min_index][edge_index] + node_d[edge_index] == node_d[min_index]:
        (x,y) = (edge_index, min_index)
        Route.append((x,y))

    del node_l[min_index]

    #3
    for j in range(len(edge[min_index].keys())):
      edge_index = list(edge[min_index].keys())[j]
      if (edge_index in node_l.keys()) and \
      (node_l[edge_index] > (edge[min_index][edge_index] + node_d[min_index])):
        node_l[edge_index] = edge[min_index][edge_index] + node_d[min_index]
```

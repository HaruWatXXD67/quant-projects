```python
import matplotlib.pyplot as plt
import numpy as np
import math
from math import sqrt
from math import exp
```

```python
x_1 = np.linspace(0,20)
y_1 = 1 + 2 * x_1
y_2 = (1 + x_1)**2

plt.plot(x_1, y_1)
plt.plot(x_1, y_2)
plt.show()
```

```python
def func(x, y):
    return 2*x**2 - 3*y**2

x = np.linspace(-np.pi * 5, np.pi * 5, 80)
y = np.linspace(-np.pi * 5, np.pi * 5, 80)
X, Y = np.meshgrid(x, y)


fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(projection='3d')

ax.plot_surface(X, Y, func(X,Y), alpha=0.7)

ax.set_xlabel('X-label')
ax.set_ylabel('Y-label')
ax.set_zlabel('Z-label')

plt.show()
```

```python
def f(x):
  return -x**3 + 3*x**2 + 9*x

x = np.linspace(-3, 6, 100)
y = f(x)

plt.plot(x, y)
plt.show()
```

```python
def df(x):
  return -3*x**2 + 6*x + 9

x_i = -3
X = [x_i]
Y = [f(x_i)]

eta = 0.01

for _ in range(10):
  x_i = x_i - eta * df(x_i)
  X.append(x_i)
  Y.append(f(x_i))

plt.plot(x, y)
plt.scatter(X,Y, color="green")
plt.show()

X[-1]
```

```python
def f(x):
  return x**3 - 2*x**2 + x + 3

x = np.linspace(0, 2, 100)
y = f(x)

plt.plot(x, y)
plt.show()
```

```python
def df(x):
  return 3*x**2 - 4*x + 1

def ddf(x):
  return 6*x - 4

x_i = 2
X = [x_i]
Y = [f(x_i)]


for _ in range(10):
  x_i = x_i - df(x_i)/ddf(x_i)
  X.append(x_i)
  Y.append(f(x_i))

plt.plot(x, y)
plt.scatter(X,Y, color="green")
plt.show()

X[-1]
```

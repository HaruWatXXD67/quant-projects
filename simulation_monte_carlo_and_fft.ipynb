```python
N = 1024
dt = 0.0005
f1, f2 = 100, 250
t = np.arange(0, N * dt, dt)
x = 1.5 * np.sin(2 * np.pi * f1 * t) + np.sin(2 * np.pi * f2 * t)

fig, ax = plt.subplots()
ax.plot(t, x)
plt.xlim(0,0.05)
ax.set_xlabel("Time [s]")
ax.set_ylabel("Signal")
ax.grid()
plt.show()
```

```python
F = np.fft.fft(x)
freq = np.fft.fftfreq(N, d=dt)

F = F / (N / 2)
Amp = np.abs(F)
fig, ax = plt.subplots()
ax.plot(freq[:N//2], Amp[:N//2])
ax.set_xlabel("Frequency [Hz]")
ax.set_ylabel("Amplitude")
ax.grid()
plt.show()
```

```python
Pow = Amp ** 2
fig, ax = plt.subplots()
ax.plot(freq[:N//2],Pow[:N//2])
ax.set_xlabel("Frequency [Hz]")
ax.set_ylabel("Power")
ax.grid()
plt.show()
```

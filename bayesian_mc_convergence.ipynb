```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)
```

```python
# Generate 1,000,000 standard normal random variables
x = np.random.randn(1000000)
n_max = len(x)

# Cumulative computations
cum_sum = np.cumsum(x)
xbar = cum_sum / np.arange(1, n_max + 1)

indicator = (x <= 0).astype(int)
cum_indicator = np.cumsum(indicator)
p_hat = cum_indicator / np.arange(1, n_max + 1)

# Log scale for n
log_n = np.log10(np.arange(1, n_max + 1))
```

```python
# Plot x̄_n
plt.figure()
plt.plot(log_n, xbar)
plt.xlabel("log10 n")
plt.ylabel("x̄_n")
plt.show()

# Plot p̂_n
plt.figure()
plt.plot(log_n, p_hat)
plt.xlabel("log10 n")
plt.ylabel("p̂_n")
plt.show()
```

```python
rng = np.random.default_rng(seed=n_max)
y = rng.normal(size=n_max)

indices = np.unique(np.geomspace(1, n_max, 1000).astype(int))

cum_medians = [np.median(y[:i]) for i in indices]
```

```python
plt.plot(np.log10(indices), cum_medians)
plt.axhline(y=0, color='red')
plt.xlabel("$\log_{10} n$")
plt.ylabel("Median Value")
plt.title(f"Convergence to True Median")
plt.show()

print(f"Final sample median at n={n_max}: {np.median(x):.6f}")
```

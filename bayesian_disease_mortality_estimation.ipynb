```python
import pandas as pd
import numpy as np
from scipy.stats import beta

data = pd.read_csv("Cholera.csv")
```

```python
n = data['popn'].values
y = data['cholera_deaths'].values

print(n)
print(y)
```

```python
#Posterior Parameters
posterior_a = y + 1
posterior_b = n - y + 1
posterior_mean = posterior_a / (posterior_a + posterior_b)
lower_95 = beta.ppf(0.025, posterior_a, posterior_b)
upper_95 = beta.ppf(0.975, posterior_a, posterior_b)

print(f"posterior_a: {posterior_a}")
print(f"posterior_b: {posterior_b}")
print(f"posterior_mean: {posterior_mean}")
print(f"lower_95_CI: {lower_95}")
print(f"upper_95_CI: {upper_95}")
```

```python
posterior_stats = pd.DataFrame({
    'district': data['district'],
    'posterior_mean': posterior_mean,
    'lower_95_CI': lower_95,
    'upper_95_CI': upper_95
})

display(posterior_stats)
```
